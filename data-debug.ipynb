```python
from pydoc import describe
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from matplotlib.patches import Rectangle
from datetime import timedelta
import logging
import mysql


"""
This is a script to analyze and visualize running balances data sourced from a pickle file.
As of 9/23/2025 this data is not sourced from a database but rather a pickle file. That file is incorrect.
"""

running_balances = pd.read_pickle('running_balances.pkl')
asset_classes = ['Certificate of Deposit', 'Mutual Fund', 'Commercial Paper', 'Money Market', 'US Treasuries', 'US Agencies']
asset_index = 3

# Sort the running balances dataframe by TransactionDate and show first 25 rows
running_balances = running_balances.sort_values('TransactionDate', ascending=True)
data = running_balances[running_balances['TransactionClass'] == asset_classes[asset_index]].copy()



# Temporarily change display options
with pd.option_context('display.max_columns', None,
                       'display.max_rows', None,
                       'display.width', None,
                       'display.expand_frame_repr', False):
    display(data.head(60))



# Display all columns with their data types
print("DataFrame columns and their types:")
print(data.dtypes)
print("\nTop rows of data:")
print(data.head(48))

```

```python
# Print basic statistics
print("\nRunning Total Statistics:")
print(data['RunningTotal'].describe())
```

```python
# Create histogram of RunningTotal
plt.figure(figsize=(10, 6))
plt.hist(data['RunningTotal'], bins=30)
plt.title(f'Distribution of Running Total for {asset_class}')
plt.xlabel('Running Total')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```
